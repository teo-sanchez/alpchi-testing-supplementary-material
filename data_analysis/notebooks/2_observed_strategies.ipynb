```python
df1 = metrics.get_final_number_of_tests_per_participant()
df2 = metrics.get_ratio_checks_added_per_participant()
df3 = metrics.get_fail_count_per_participant()
df4 = metrics.get_gini_impurity_per_participant()


# Merge on Label
df = df1.merge(df2, on="participant_id").merge(df3, on="participant_id").merge(df4, on="participant_id")
print(df)
# Define the metrics to compare
metrics_name = ['Total test cases', 'Ratio checks/added', 'Failed test cases', 'Gini impurity']

# Compute and store correlations and p-values
results = []
for i in range(len(metrics_name)):
    for j in range(i + 1, len(metrics_name)):
        metric_x = metrics_name[i]
        metric_y = metrics_name[j]
        corr, p_value = spearmanr(df[metric_x], df[metric_y])
        results.append({
            "Metric X": metric_x,
            "Metric Y": metric_y,
            "Correlation": round(corr, 2),
            "p-value": round(p_value, 5)
        })

# Convert results to a DataFrame for display
results_df = pd.DataFrame(results)
print(results_df)
# Plot correlation between Fail ratio and Err. rate using seaborn
```
